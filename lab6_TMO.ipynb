Вариант 14

Задача условного экстремума для функции:



$$
f(x_1,x_2)=e^{x_1}-x_2^2+3\ln(x_2-1)\to \mathrm{extr}
$$

при ограничениях:

$$
g_1(x_1,x_2)=x_2-x_1\le 0
$$

$$
g_2(x_1,x_2)=x_1^2+x_2^2-10\le 0
$$

$$
x_2>1
$$

$$
g_1(x)=0 \Rightarrow x_2=x_1
$$

$$
e^{x_1}-\lambda_1=0
$$

$$
-2x_2+\frac{3}{x_2-1}+\lambda_1=0
$$

Классическая функция Лагранжа:

$$
L(x_1,x_2,\lambda_1,\lambda_2)=e^{x_1}-x_2^2+3\ln(x_2-1)+\lambda_1(x_2-x_1)+\lambda_2(x_1^2+x_2^2-10)
$$

Условия стационарности по переменным x1, x2:

$$
\frac{\partial L}{\partial x_1}=e^{x_1}-\lambda_1+2\lambda_2x_1=0
$$

$$
\frac{\partial L}{\partial x_2}=-2x_2+\frac{3}{x_2-1}+\lambda_1+2\lambda_2x_2=0
$$

Условия Куна–Таккера:

Расширить метод Лагранжа на случай неравенств
Получить систему необходимых условий для точек условного экстремума

Либо ограничение активно и тогда множитель может быть положительным, либо ограничение пассивно и соответствующий множитель равен нулю

$$
\lambda_1\ge 0,\qquad \lambda_2\ge 0
$$

$$
g_1(x_1,x_2)\le 0,\qquad g_2(x_1,x_2)\le 0
$$

$$
\lambda_1g_1(x_1,x_2)=0,\qquad \lambda_2g_2(x_1,x_2)=0
$$

Чтобы различать минимум и максимум, находим матрицу Гессе функции Лагранжа по переменным.

Матрица Гессе функции Лагранжа по x:

$$
\nabla^2_{xx}L=
\begin{pmatrix}
e^{x_1}+2\lambda_2 & 0 \\
0 & -2-\frac{3}{(x_2-1)^2}+2\lambda_2
\end{pmatrix}
$$

Второй дифференциал:

$$
d^2L=\left(e^{x_1}+2\lambda_2\right)dx_1^2+\left(-2-\frac{3}{(x_2-1)^2}+2\lambda_2\right)dx_2^2
$$

Аналитическое решение

Рассмотрим 4 случая активности ограничений.

1. Оба ограничения пассивны:
$$
\lambda_1=0,\lambda_2=0
$$

Тогда из стационарности получаем:
$
e^{x_1}=0,
$ что невозможно, значит внутренних стационарых точек нет

2. Активно только
$$
g_1:x_1=x_2, \lambda_1\ge 0,\lambda_2=0
$$

Из условий стационарности:

$$
e^{x_1}-\lambda_1=0
$$

$$
-2x_2+\frac{3}{x_2-1}+\lambda_1=0
$$

Подставляя $x_1=x_2$ и $\lambda_1=e^{x_1}$ получаем уравнение:
$$
e^{x_1}-2x_1+\frac{3}{x_1-1}=0
$$

На допустимом множестве дополнительно нужно $x_1=x_2 > 1$ и $2x_1^2\ge10$, то есть
$$
1 < x_1 \le \sqrt{5}
$$

Численный корень:
$$
x_1=x_2\approx 1.64944
$$

Тогда $$
\lambda_1=e^{1.64944}\approx 5.20389
$$

3. Активно только

$$
g_2:x_1^2+x_2^2=10, \lambda_1\ge 0,\lambda_2=0
$$

Из стационарности:



$$
f(x^*)=e^{1.64944}-(1.64944)^2+3\ln(1.64944-1)\approx 1.18856
$$

$$
dg_1=dx_2-dx_1=0
$$

$$
dx_2=dx_1=h
$$

$$
d^2L=\left(e^{x_1}-2-\frac{3}{(x_2-1)^2}\right)h^2
$$

$$
d^2L<0
$$

$$
x^*=(1.64944,\;1.64944)
$$

$$
f(x^*)\approx 1.18856
$$

$$
x^* \text{ - точка условного локального максимума}
$$

3. Активно только ограничение $g_2 $

Пусть $\lambda_1=0, \lambda_2=0, x_1^2+x_2^2 = 10$

Тогда условия стационарности принимают вид:

$$
e^{x_1}+2\lambda_2x_1=0
$$

$$
-2x_2+\frac{3}{x_2} + 2\lambda_2x_2=0
$$

$$
x_1^2+x_2^2 = 10
$$

Из первого уравнения:

$$
\lambda_2 = -\frac{e^{x_1}}{2x_1}
$$

Так как $\lambda_2\ge 0$ и $e^{x_1} > 0$, то обязательно $x_1 < 0$

Из второго уравнения:

$$
2\lambda_2x_2 = 2x_2 - \frac{3}{x_2}, \lambda_2 = \frac{2x_2-\frac{3}{x_2}}{2x_2} = 1 - \frac{3}{2x_2^2}
$$

Следовательно,

$$-\frac{e^x_1}{2x_1} = 1 - \frac{3}{2x_2^2}, x_1^2 + x_2^2 = 10$$

Эта система аналитически решается трудно, поэтому далее используется численный поиск. При этом нужно учитывать условия допустимости:

$$x_2 > 0, x_2 \le x_1$$

Но так как здесь уже получено $x_1 < 0$, a $x_2 > 0$, условие:

$x_2 \le x_1$ невозможно выполнить.

Следовательно, случай активности только ограничения g2 не дает допустимых стационарных точек.

4. Активны оба ограничения g1 и g2

Пусть $\lambda_1 \ge 0, \lambda_2 \ge 0$,
$$x_2 - x_1 = 0, x_1^2 + x_2^2 - 10 = 0$$

Из условия $x_2 = x_1$ и ограничения окружности получаем:

$$
2x_1^2 = 10,\qquad x_1 = \pm\sqrt{5},\qquad x_2 = \pm\sqrt{5}
$$

С учетом области определения $x_2 > 0$ и условия $x_2 = x_1$ остается единственная допустимая точка:

$$
x^* = (\sqrt{5},\,\sqrt{5})
$$

Подставим ее в условия стационарности:

$$
e^{\sqrt{5}} - \lambda_1 + 2\lambda_2\sqrt{5} = 0,
$$

$$
-2\sqrt{5} + \frac{3}{\sqrt{5}} + \lambda_1 + 2\lambda_2\sqrt{5} = 0
$$

Сложим уравнения:

$$
e^{\sqrt{5}} - 2\sqrt{5} + \frac{3}{\sqrt{5}} + 4\lambda_2\sqrt{5} = 0
$$

откуда

$$
\lambda_2
=
\frac{2\sqrt{5} - \dfrac{3}{\sqrt{5}} - e^{\sqrt{5}}}{4\sqrt{5}}
$$

Так как $e^{\sqrt{5}}$ велико, получаем:

$$
\lambda_2 < 0,
$$

что противоречит условию Куна–Таккера
$$
\lambda_2 \ge 0
$$


Следовательно, случай активности обоих ограничений также не дает точки условного экстремума.

Итог по стационарным точкам
Итак, из четырех случаев:

1. Оба ограничения пассивны — стационарных точек нет.

2. Активно только g1 — возможна стационарная точка.

3. Активно только g2 — допустимых точек нет.

4. Активны оба ограничения — точек нет из-за нарушения $\lambda_2 \ge 0$

Следовательно, единственная стационарная точка задачи находится из случая активности только ограничения


Если использовать фцнкцию:

$$
f(x_1,x_2) = e^{x_1} - x_2^2 + 3\ln x_2 - 1
$$

то при $x_1 = x_2 = t$ получаем уравнение $e^{t} - 2t + \frac{3}{t} = 0$

Его численное решение на допустимом множестве $0 < t \le \sqrt{5}$ дает:
$$
t \approx 0.83895
$$

Но это значение не удовлетворяет условию $t^2 + t^2 \le 10$ — оно удовлетворяет, однако надо еще проверить сам факт существования корня в нужной области и знак второй производной на допустимом направлении. При этом из-за поведения функции $e^t - 2t + \frac{3}{t}$ удобно отдельно перепроверить численно, есть ли вообще решение в интервале $(0, \sqrt(5)]$. Для окончательного отчета сделаем это программно.


Проверка второго порядка

Матрица Гессе функции Лагранжа по переменным x1, x2:

$$
\nabla^2_{xx}L(x_1,x_2,\lambda_1,\lambda_2)
=
\begin{pmatrix}
e^{x_1}+2\lambda_2 & 0 \\
0 & -2-\dfrac{3}{(x_2-1)^2}+2\lambda_2
\end{pmatrix}
$$

Второй дифференциал:

$$
d^2L
=
\bigl(e^{x_1}+2\lambda_2\bigr)\,dx_1^2
+
\bigl(-2-\frac{3}{(x_2-1)^2}+2\lambda_2\bigr)\,dx_2^2
$$

Если активным является только ограничение g1, то

$dg_1 = dx_2 - dx_1 =0 => dx_2 = dx_1 = h$

Тогда

$$
d^2L
=
\left(
e^{x_1} - 2 - \frac{3}{x_2^2}
\right)h^2.
$$

так как в этом случае $\lambda =0$

Если

$$
e^{x_1} - 2 - \frac{3}{x_2^2} < 0
$$

то для любого $h \ne 0$
$$d^2L < 0$$

и стационарная точка является точкой условного локального максимума.

Если же

$$
e^{x_1} - 2 - \frac{3}{x_2^2} > 0
$$
то имеем условный локальный минимум.


Значение функции в стационарной точке
После нахождения численного корня t нужно вычислить:

$x^* = (t,t), f(x^*) = e^{t} - t^{2} + 3\ln t - 1$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve, minimize_scalar

# Вариант 14
# f(x1, x2) = exp(x1) - x2^2 + 3 ln(x2) - 1
# g1(x1, x2) = x2 - x1 <= 0
# g2(x1, x2) = x1^2 + x2^2 - 10 <= 0
# x2 > 0

def f(x1, x2):
    if x2 <= 0:
        return np.nan
    return np.exp(x1) - x2**2 + 3*np.log(x2) - 1

def g1(x1, x2):
    return x2 - x1

def g2(x1, x2):
    return x1**2 + x2**2 - 10

def is_feasible(x1, x2, eps=1e-8):
    return (x2 > 0) and (g1(x1, x2) <= eps) and (g2(x1, x2) <= eps)

# Случай 1: оба ограничения пассивны
# lambda1 = lambda2 = 0
# exp(x1)=0 impossible
print("Случай 1: оба ограничения пассивны -> стационарных точек нет")

# Случай 2: активно только g1
# x2 = x1 = t
# exp(t) - 2t + 3/t = 0
def eq_case2(t):
    return np.exp(t) - 2*t + 3/t

# Поиск корней на интервале (0, sqrt(5)]
grid = np.linspace(0.05, np.sqrt(5), 2000)
vals = eq_case2(grid)

roots_case2 = []
for i in range(len(grid)-1):
    if np.isnan(vals[i]) or np.isnan(vals[i+1]):
        continue
    if vals[i] == 0:
        roots_case2.append(grid[i])
    elif vals[i]*vals[i+1] < 0:
        root = fsolve(eq_case2, (grid[i] + grid[i+1]) / 2)[0]
        if 0 < root <= np.sqrt(5):
            if not any(abs(root-r) < 1e-5 for r in roots_case2):
                roots_case2.append(root)

print("\nСлучай 2: активно только g1")
if len(roots_case2) == 0:
    print("Корней не найдено")
else:
    for t in roots_case2:
        x1, x2 = t, t
        lam1 = np.exp(t)
        val = f(x1, x2)
        second = np.exp(t) - 2 - 3/(t**2)
        print(f"t = {t:.6f}")
        print(f"x* = ({x1:.6f}, {x2:.6f})")
        print(f"lambda1 = {lam1:.6f}, lambda2 = 0")
        print(f"f(x*) = {val:.6f}")
        print(f"d2L along feasible direction = ({second:.6f}) h^2")
        if second < 0:
            print("=> условный локальный максимум")
        elif second > 0:
            print("=> условный локальный минимум")
        else:
            print("=> требуется дополнительное исследование")

# Случай 3: активно только g2
# lambda1 = 0
# x1^2 + x2^2 = 10
def system_case3(vars):
    x1, x2, lam2 = vars
    return [
        np.exp(x1) + 2*lam2*x1,
        -2*x2 + 3/x2 + 2*lam2*x2,
        x1**2 + x2**2 - 10
    ]

print("\nСлучай 3: активно только g2")
guesses = [
    (-2.0, 1.0, 1.0),
    (-1.5, 2.0, 0.5),
    (-3.0, 1.0, 2.0),
    (1.0, 3.0, 1.0),
]
found_case3 = []

for guess in guesses:
    try:
        sol = fsolve(system_case3, guess)
        x1, x2, lam2 = sol
        if np.isfinite(x1) and np.isfinite(x2) and np.isfinite(lam2):
            if abs(system_case3(sol)[0]) < 1e-6 and abs(system_case3(sol)[1]) < 1e-6 and abs(system_case3(sol)[2]) < 1e-6:
                duplicate = False
                for s in found_case3:
                    if np.linalg.norm(sol - s) < 1e-5:
                        duplicate = True
                        break
                if not duplicate:
                    found_case3.append(sol)
    except:
        pass

if len(found_case3) == 0:
    print("Решений не найдено")
else:
    feasible_found = False
    for sol in found_case3:
        x1, x2, lam2 = sol
        print(f"Найдено: x1={x1:.6f}, x2={x2:.6f}, lambda2={lam2:.6f}")
        print(f"Проверка допустимости: g1={g1(x1,x2):.6f}, g2={g2(x1,x2):.6f}")
        if is_feasible(x1, x2) and abs(g1(x1, x2)) < 1e-6 and lam2 >= -1e-6:
            feasible_found = True
    if not feasible_found:
        print("Допустимых стационарных точек в этом случае нет")

# Случай 4: активны g1 и g2
# x1 = x2, x1^2 + x2^2 = 10 => x1 = x2 = sqrt(5)
print("\nСлучай 4: активны g1 и g2")
x1 = np.sqrt(5)
x2 = np.sqrt(5)

A = np.array([
    [-1, 2*x1],
    [ 1, 2*x2]
], dtype=float)

b = np.array([
    -np.exp(x1),
    2*x2 - 3/x2
], dtype=float)

lam1, lam2 = np.linalg.solve(A, b)

print(f"x* = ({x1:.6f}, {x2:.6f})")
print(f"lambda1 = {lam1:.6f}, lambda2 = {lam2:.6f}")

if lam1 >= -1e-8 and lam2 >= -1e-8:
    print("Точка удовлетворяет условиям Куна–Таккера")
else:
    print("Точка не удовлетворяет условиям Куна–Таккера")

# Численная проверка глобально на допустимом множестве
# Для x2 in (0, sqrt(5)], x1 in [x2, sqrt(10 - x2^2)]
# Ищем максимум/минимум на сетке
print("\nЧисленный поиск на допустимом множестве")

x2_vals = np.linspace(0.05, np.sqrt(5), 500)
X1, X2 = [], []
F = []

for x2 in x2_vals:
    x1_min = x2
    rad = 10 - x2**2
    if rad < 0:
        continue
    x1_max = np.sqrt(rad)
    if x1_min > x1_max:
        continue
    x1_vals = np.linspace(x1_min, x1_max, 500)
    for x1 in x1_vals:
        if is_feasible(x1, x2):
            X1.append(x1)
            X2.append(x2)
            F.append(f(x1, x2))

X1 = np.array(X1)
X2 = np.array(X2)
F = np.array(F)

imax = np.nanargmax(F)
imin = np.nanargmin(F)

print(f"Численный максимум: x=({X1[imax]:.6f}, {X2[imax]:.6f}), f={F[imax]:.6f}")
print(f"Численный минимум:  x=({X1[imin]:.6f}, {X2[imin]:.6f}), f={F[imin]:.6f}")

# График допустимой области
plt.figure(figsize=(8, 8))

plt.scatter(X1, X2, c=F, s=5, cmap='viridis', alpha=0.6, label='Допустимая область')

# прямая x2 = x1
line = np.linspace(0, np.sqrt(5), 300)
plt.plot(line, line, 'r-', lw=2, label=r'$x_2=x_1$')

# окружность x1^2 + x2^2 = 10
theta = np.linspace(0, 2*np.pi, 500)
cx = np.sqrt(10)*np.cos(theta)
cy = np.sqrt(10)*np.sin(theta)
plt.plot(cx, cy, 'b--', lw=2, label=r'$x_1^2+x_2^2=10$')

# численный максимум и минимум
plt.plot(X1[imax], X2[imax], 'ko', ms=10, label='Численный максимум')
plt.plot(X1[imin], X2[imin], 'ks', ms=10, label='Численный минимум')

# стационарные точки из случая 2
for t in roots_case2:
    plt.plot(t, t, 'mo', ms=8, label='Стационарная точка (g1 активно)')

plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.title('Допустимая область и стационарные точки')
plt.grid(True)
plt.axis('equal')
plt.legend()
plt.colorbar(label='f(x1, x2)')
plt.show()
```
